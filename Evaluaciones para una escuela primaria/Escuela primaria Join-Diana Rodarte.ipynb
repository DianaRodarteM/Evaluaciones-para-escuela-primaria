```python
import sqlite3

# Conectar o crear base de datos
conn = sqlite3.connect("escuela_primaria.db")
cursor = conn.cursor()

# Eliminar tablas si ya existen
cursor.execute("DROP TABLE IF EXISTS calificaciones")
cursor.execute("DROP TABLE IF EXISTS alumnos")
cursor.execute("DROP TABLE IF EXISTS materias")
cursor.execute("DROP TABLE IF EXISTS grupos")

# Crear tabla grupos
cursor.execute("""
CREATE TABLE grupos (
    id_grupo INTEGER PRIMARY KEY,
    nombre_grupo TEXT NOT NULL,
    grado INTEGER NOT NULL,
    turno TEXT NOT NULL
)
""")

# Crear tabla alumnos
cursor.execute("""
CREATE TABLE alumnos (
    id_alumno INTEGER PRIMARY KEY,
    nombre TEXT NOT NULL,
    apellido TEXT NOT NULL,
    edad INTEGER NOT NULL,
    id_grupo INTEGER,
    FOREIGN KEY (id_grupo) REFERENCES grupos(id_grupo)
)
""")

# Crear tabla materias
cursor.execute("""
CREATE TABLE materias (
    id_materia INTEGER PRIMARY KEY,
    nombre_materia TEXT NOT NULL
)
""")

# Crear tabla calificaciones
cursor.execute("""
CREATE TABLE calificaciones (
    id_calificacion INTEGER PRIMARY KEY,
    id_alumno INTEGER,
    id_materia INTEGER,
    calificacion REAL,
    FOREIGN KEY (id_alumno) REFERENCES alumnos(id_alumno),
    FOREIGN KEY (id_materia) REFERENCES materias(id_materia)
)
""")

conn.commit()
print("Base de datos y tablas creadas correctamente.")
```

```python
grupos.to_sql("grupos", conn, if_exists="append", index=False)
alumnos.to_sql("alumnos", conn, if_exists="append", index=False)
materias.to_sql("materias", conn, if_exists="append", index=False)
calificaciones.to_sql("calificaciones", conn, if_exists="append", index=False)

print("Datos insertados en SQLite correctamente.")
```

```python
query = """
SELECT alumnos.id_alumno,
       alumnos.nombre,
       alumnos.apellido,
       grupos.nombre_grupo,
       grupos.grado,
       grupos.turno
FROM alumnos
INNER JOIN grupos ON alumnos.id_grupo = grupos.id_grupo;
"""

df = pd.read_sql_query(query, conn)
df
```

```python
query = """
SELECT alumnos.id_alumno,
       alumnos.nombre,
       alumnos.apellido,
       calificaciones.calificacion
FROM alumnos
LEFT JOIN calificaciones ON alumnos.id_alumno = calificaciones.id_alumno;
"""

df = pd.read_sql_query(query, conn)
df
```

```python
query = """
SELECT alumnos.nombre,
       alumnos.apellido,
       materias.nombre_materia,
       calificaciones.calificacion,
       CASE
            WHEN calificaciones.calificacion >= 6 THEN 'Aprobado'
            ELSE 'Reprobado'
       END AS estado
FROM calificaciones
INNER JOIN alumnos ON calificaciones.id_alumno = alumnos.id_alumno
INNER JOIN materias ON calificaciones.id_materia = materias.id_materia;
"""

df = pd.read_sql_query(query, conn)
df
```

```python
query = """
SELECT grupos.nombre_grupo,
       grupos.grado,
       ROUND(AVG(sub.promedio_alumno), 2) AS promedio_grupo
FROM grupos
INNER JOIN alumnos ON grupos.id_grupo = alumnos.id_grupo
INNER JOIN (
    SELECT id_alumno,
           AVG(calificacion) AS promedio_alumno
    FROM calificaciones
    GROUP BY id_alumno
) sub ON alumnos.id_alumno = sub.id_alumno
GROUP BY grupos.id_grupo;
"""

df = pd.read_sql_query(query, conn)
df
```

```python
query = """
SELECT *
FROM alumnos
WHERE id_alumno IN (
    SELECT id_alumno
    FROM calificaciones
);
"""

df = pd.read_sql_query(query, conn)
df
```

```python
query = """
SELECT *
FROM alumnos
WHERE id_alumno NOT IN (
    SELECT id_alumno
    FROM calificaciones
);
"""

df = pd.read_sql_query(query, conn)
df
```
